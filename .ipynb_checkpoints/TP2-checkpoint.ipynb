```python
import seaborn as sns
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
```

```python
titanic = sns.load_dataset('titanic')
print('Dimensions :', titanic.shape)
display(titanic.head())
```

```python
print('Colonnes :', titanic.columns.tolist())
print('\nTypes :')
print(titanic.dtypes)
```

```python
qualitatives = titanic.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()
quantitatives = titanic.select_dtypes(include=['number']).columns.tolist()
print('Variables qualitatives :', qualitatives)
print('Variables quantitatives :', quantitatives)
```

```python
for col in qualitatives:
    print(f'\nRepartition de {col}:')
    print(titanic[col].value_counts(dropna=False))
```

```python
display(titanic[quantitatives].describe())
```

```python
missing_per_col = titanic.isnull().sum()
print('Valeurs manquantes par colonne :')
print(missing_per_col)
print('\nTotal des valeurs manquantes :', missing_per_col.sum())
```

```python
df1 = titanic.dropna()
df2 = titanic.drop(columns=['deck']).dropna()
print('Taille de df1 :', df1.shape)
print('Taille de df2 :', df2.shape)
```

```python
df3 = titanic[['age', 'fare']].copy()
df3['age'] = df3['age'].fillna(df3['age'].median())
print(df3.isnull().sum())
display(df3.head())
```

```python
scaler = MinMaxScaler()
df3_norm = pd.DataFrame(scaler.fit_transform(df3), columns=df3.columns)
display(df3_norm.head())
display(df3_norm.describe())
```

```python

df_flights = sns.load_dataset('flights')
print('Avant encodage :')
display(df_flights.head())
```

```python
try:
    encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
except TypeError:
    encoder = OneHotEncoder(sparse=False, handle_unknown='ignore')

month_encoded = encoder.fit_transform(df_flights[['month']])
encoded_columns = encoder.get_feature_names_out(['month'])
encoded_df = pd.DataFrame(month_encoded, columns=encoded_columns, index=df_flights.index)
df_encoded = pd.concat([df_flights.drop(columns=['month']), encoded_df], axis=1)

print('Apres encodage :')
display(df_encoded.head())
```
