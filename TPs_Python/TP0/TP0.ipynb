```python
import numpy as np
import matplotlib.pyplot as plt


# --------------------------
# Partie I : Variables, listes, tuples
# --------------------------
a = 2
b = 2.0
c = "Hello Python"
u = True
z = 3 + 5.2j

print(a, type(a))
print(b, type(b))
print(c, type(c))
print(u, type(u))
print(z, type(z))

x = 5
y = 3
print("x+y=", x + y)
print("x-y=", x - y)
print("x/y=", x / y)
print("x*y=", x * y)
print("x**y=", x**y)
print("x%y=", x % y)
print("x//y=", x // y)

l = []
print(l)

L = [1, 2.0, True, "Hello", [1, "g"]]
print(L)

L1 = ["souris", "girafe"]
L2 = [1.0, 4.32, 0, True, False]
print(L1 + L2)
print(3 * L1)

L1.append([0, "Merci"])
L2.insert(3, "Bonsoir")
del L2[3]


# --------------------------
# Partie II : Fonctions
# --------------------------
def somme(a_val, b_val):
    return a_val + b_val


print("somme(12,17) =", somme(12, 17))

prod = lambda u_val, v_val: u_val * v_val - 2 * v_val


def calcul(x_val, y_val):
    p = prod(x_val, y_val)
    if p == 0:
        return None
    return 1 / p


print("prod(4,3) =", prod(4, 3))
print("calcul(4,3) =", calcul(4, 3))


def comparaison(a_val, b_val):
    if a_val > b_val:
        return "a est plus grand que b"
    if a_val < b_val:
        return "a est plus petit que b"
    return "a est egale a b"


def somme_rec(n):
    if n == 0:
        return 0
    return n + somme_rec(n - 1)


def parite(values):
    subset = values[7:16]
    return [f"{v} est pair" if v % 2 == 0 else f"{v} est impair" for v in subset]


L_range = [i for i in range(2, 20)]
print(parite(L_range))


def factoriel(n):
    i = 1
    f = 1
    while i <= n:
        f *= i
        i += 1
    return f


print("5! =", factoriel(5))


# --------------------------
# Partie IV : Matrices
# --------------------------
A = np.array([[4, 3, 3, 3], [2, 4, 3, 3], [2, 2, 4, np.pi], [2, 2, 2, 0.5]])
print(A, type(A))

print("A[0,0] =", A[0, 0], ", A[3,3] =", A[3, 3], ", A[2,3] =", A[2, 3])

V = A[:, 1:2]
W = A[:, 1:3]
print("V=\n", V)
print("W=\n", W)

B = A.copy()
B[2, 3] = np.e
B[3, :] = [1, 1, 1, 1]
print("B=\n", B)

print("diag(A):", np.diag(A))
print("sous diag k=-1:", np.diag(A, k=-1))
print("sur diag k=2:", np.diag(A, k=2))

print("tril(A)=\n", np.tril(A))
print("triu(A)=\n", np.triu(A))
print("A@V=\n", A @ V)
print("A^5=\n", np.linalg.matrix_power(A, 5))
print("A.T=\n", A.T)

U_vec = [-2, 1, 2]
C = np.diagflat(U_vec)
D = np.diag(np.diag(A))
print("C=\n", C)
print("D=\n", D)

M = np.array([[4, -1, -1, 0], [-1, 4, 0, -1], [-1, 0, 4, -1], [0, -1, -1, 4]])
b_vec = np.array([[50], [30], [70], [50]])
X_sol = np.linalg.solve(M, b_vec)
print("Solution du systeme:\n", X_sol)


def def_mat(n):
    mat = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            if i > j:
                mat[i, j] = (i + 1) + (j + 1) + 1
            elif i < j:
                mat[i, j] = (i + 1) + (j + 1) - 1
            else:
                mat[i, j] = i + 1
    return mat


print("def_mat(3)=\n", def_mat(3))


# --------------------------
# Partie V : Approximation de pi
# --------------------------
def wallis_pi(n):
    p = 1.0
    for k in range(1, n + 1):
        p *= (4 * k * k) / (4 * k * k - 1)
    return 2 * p


def euler_pi(n):
    s = 0.0
    for k in range(1, n + 1):
        s += 1 / (k**2)
    return np.sqrt(6 * s)


n_values = range(10, 1001, 10)
errors_wallis = [abs((np.pi - wallis_pi(n)) / wallis_pi(n)) for n in n_values]
errors_euler = [abs((np.pi - euler_pi(n)) / euler_pi(n)) for n in n_values]

plt.plot(n_values, errors_wallis, label="Wallis")
plt.plot(n_values, errors_euler, label="Euler")
plt.xlabel("Nombre d'iterations (n)")
plt.ylabel("Erreur relative")
plt.title("Evolution de l'erreur relative")
plt.grid(True)
plt.legend()
plt.show()
```
